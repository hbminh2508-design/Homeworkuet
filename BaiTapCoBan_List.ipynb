### Bài 2: Tìm số nhỏ nhất của một danh sách các số nguyên. KHONG DUNG HAM SAN CO CUA LIST

```python
lst = list(map(int, input().split()))

def min_of_list (danh_sach):
  min_val = danh_sach[0]
  for i in range(len(danh_sach)):
    if min_val > danh_sach[i]:
      min_val = danh_sach[i]

  return min_val

print(min_of_list(lst))
```

### Bài 2b: Tìm số nhỏ nhất của 1 list và vị trí của nó, nếu có nhiều số nhỏ nhất bằng nhau thì lấy chỉ số đầu tiên (nhỏ nhất)
10 0 0 12 34 5 6 7  8 89 0 1
0 1

```python
lst = list(map(int, input().split()))

def min_and_first_in_list(danh_sach):
  min_val = danh_sach[0]
  vi_tri = 0
  for i in range(len(danh_sach)):
    if min_val > danh_sach[i]:
      min_val = danh_sach[i]
      vi_tri = i

      if min_val == danh_sach[i]:
        break

  return min_val, vi_tri

print(min_and_first_in_list(lst))
```

### Bài 3:
Tìm giá trị trung bình của các số nguyên dương trong một list.
Nếu không có số dương nào thì trả về "invalid"

```python
lst = list(map(int, input().split()))

def Average_positive_numbers_of_list(danh_sach):
  tong = 0
  dem = 0
  for i in range(len(danh_sach)):
    if danh_sach[i] > 0:
      tong += danh_sach[i]
      dem += 1

  if dem == 0:
    return "invalid"
  else:
    return tong / dem

print(Average_positive_numbers_of_list(lst))
```

### Bài 4: Kiểm tra một list số nguyên có phải là đối xứng hay không

### Bài 5: Đưa tất cả các số bằng 0 trong list về đầu của list. Các số khác không thay đổi thứ tự trước sau
3 4 6 7 1 9 0 0 23 0 0 34 67 89
0 0 0 0 3 4 6 7 1 9 23 34 67 89

```python
def move_zero_to_first(danh_sach):
  zero = []
  other = []

  for i in range(len(danh_sach)):
    if danh_sach[i] == 0:
      zero.append(danh_sach[i])
    else:
      other.append(danh_sach[i])

  return zero + other

lst = list(map(int, input().split()))
print(*move_zero_to_first(lst))
```

### Bài 6: Xóa tất cả các số giống nhau trong list, giữ lại đúng 1 lần, Các số khác giữ nguyên vị trí trước
[1, 2, 3, 4, 1, 2, 3, 4, 1, 1, 4, 3]


```python
def unique_array(danh_sach):

  so_duynhat = []

  for x in danh_sach:
    if x not in so_duynhat:
      so_duynhat.append(x)

  return so_duynhat

lst = list(map(int, input().split()))
print(unique_array(lst))
```

### Bài 7: Kiểm tra một list số nguyên có phải là dãy tăng, dãy giảm hay vô danh
1 3 5 6 7 8 9: Tang

9 8 7 6 5 4 0: Giam

1 2 3 7 1 0 2: Vo danh

1 2 2 4 5 7: Vo danh

```python
def check_order_of_list(danh_sach):
  is_tang = True
  is_giam = True
  for i in range(1, len(danh_sach)):
    if danh_sach[i - 1] < danh_sach[i]:
      is_giam = False
    elif danh_sach[i - 1] > danh_sach[i]:
      is_tang = False
    else:
      is_tang = False
      is_giam = False

  if is_tang == True:
    return "Tang"
  elif is_giam == True:
    return "Giam"
  else:
    return "Vo danh"
lst = list(map(int, input().split()))
print(check_order_of_list(lst))
```

### Bài 8: Kiểm tra một danh sách số nguyên có phải là ziczac khong
1 3 2 5 2 8 7 9 2 100: ziczac
34 2 56 7 90 20 30 7 120: ziczac
2 2 34 4 56 7 89 : không phải ziczac


```python
def is_ziczac(danh_sach):
  is_tang = True
  is_giam = True

  for i in range(2, len(danh_sach)):

    la_dinh = danh_sach[i - 2] < danh_sach[i - 1] > danh_sach[i]
    la_bung = danh_sach[i - 2] > danh_sach[i - 1] < danh_sach[i]

    if not (la_dinh or la_bung):
      return False

  return True

lst = list(map(int, input().split()))
print(is_ziczac(lst))
```

### Bài 9: Tìm và in ra tất cả các vị trí của số nguyên k trong một danh sách số nguyên cho trước. Nếu không có k trong ds thì in ra "-1".
VD: k = 3, [1,3,4,4,3,3,4,3,6,1] => k xuất hiện tại các vị trí (trả về 1 list) [1, 4, 5, 7]
k = 10, [1,3,4,4,3,3,4,3,6,1] => k không xuất hiện trong ds => trả về None

```python
def find_pos_of_k(lst, k):
  vi_tri = []
  for i in range(len(lst)):
    if int(lst[i]) == k:
      vi_tri.append(i)
  return vi_tri

k = int(input())
danh_sach = input().split()

print(find_pos_of_k(danh_sach, k))
```

### Bài 11:
Cho một danh sách các số nguyên có giá trị trong đoạn [0,9]. Liệt kê số lần xuất hiện của các số có mặt trong danh sách. Số nào xuất hiện trước thì liệt kê trước
VD 1 2 1 2 1 1 1 1

output:
1: 6
2: 2


```python
def list_number_in_lst(lst):
  dem = {}
  for so in lst:
    if int(so) > 9:
      return None

    else:
      if so in dem:
        dem[so] += 1
      else:
        dem[so] = 1
  return dem


danh_sach = input().split()
print(list_number_in_lst(danh_sach))
```

### Bài 11B:
Cho một danh sách các số nguyên có giá trị trong đoạn [0,9]. Liệt kê số lần xuất hiện của các số có mặt trong danh sách. Số nào nhỏ thì liệt kê trước
VD 2 1 2 1 2 1 1 1 1

output: dict



```python
def list_number_in_lst_order(lst):
  dem = {}
  for so in lst:
    so_int = int(so)
    if so_int > 9:
      return None
    else:
      if so_int in dem:
        dem[so_int] += 1
      else:
        dem[so_int] = 1
  sorted_dem = dict(sorted(dem.items()))
  return sorted_dem

danh_sach = input().split()
print(list_number_in_lst_order(danh_sach))
```

### Bài 12:
Cho một chuỗi kí tự chỉ gồm các kí tự chữ cái thường 'a'....'z'. Hãy liệt kê số lần xuất hiện của các kí tự có mặt trong chuỗi theo thứ tự tăng dần.
VD: "abcabcccc"
output:
a : 2
b : 2
c : 5

```python
def count_character_of_string(s):
  dem = {}

  if not s.isalpha():
    return None

  else:
    for char in s:
      if char in dem:
        dem[char] += 1
      else:
        dem[char] = 1

  sorted_char = sorted(dem.items())
  for char, so_luong in sorted_char:
    print(f"{char} : {so_luong}", end = " ")

danh_sach = input()
count_character_of_string(danh_sach)
```
