### Bài W8A16
**Yêu cầu:** Xây dựng lớp Shape với phương thức area(). Lớp Rectangle(w, h) và Circle(r) kế thừa, ghi đè area(). Lấy pi=3.14. In ra diện tích với 2 chữ số thập phân.

```python
class Shape:
  def area(self):
    return 0

class Rectangle:
  def __init__(self, w, h):
    self.w = w
    self.h = h

  def area(self):
    return self.w * self.h

class Circle:
  def __init__(self, r):
    self.r = r

  def area(self):
    pi = 3.14
    return self.r * pi * self.r

#-----NHẬP-CÁC-CHỈ-SỐ-----#
nhap_hcn = input().split()
w = float(nhap_hcn[0])
h = float(nhap_hcn[1])

r = float(input())

hcn = Rectangle(w, h)
ht = Circle(r)

print(f"Diện tích hình chữ nhật là: {hcn.area():.2f}")
print(f"Diện tích hình tròn là: {ht.area():.2f}")
```

**Yêu cầu:** Xây dựng lớp Vehicle(brand, speed) và Car(brand, speed, seats) kế thừa từ Vehicle. Viết phương thức info() cho mỗi lớp.

```python
class Vehicle:
  def __init__(self, brand, speed):
    self.brand = brand
    self.speed = speed

  def info(self):
    return f"{self.brand}-{self.speed}"

class Car(Vehicle):
  def __init__(self, brand, speed, seats):
    super().__init__(brand, speed)
    self.seats = seats

  def info(self):
    return f"{self.brand}-{self.speed}-{self.seats}"

loai_doi_tuong = input().strip()
du_lieu = input().strip().split()

if loai_doi_tuong == "Vehicle":
  brand = du_lieu[0]
  speed = du_lieu[1]

  xe = Vehicle(brand, speed)
  print(xe.info())

elif loai_doi_tuong == "Car":
  brand = du_lieu[0]
  speed = du_lieu[1]
  seats = du_lieu[2]

  oto = Car(brand, speed, seats)
  print(oto.info())
```

### Bài W8A20
**Yêu cầu:** Xây dựng lớp Account(balance) có deposit(x), get_balance(). Lớp SavingAccount(balance, rate) kế thừa, thêm add_interest() để cộng balance*rate vào số dư.

```python
class Account:
  def __init__(self, balance):
    self.balance = float(balance)

  def deposit(self, x):
    self.balance += float(x)

  def get_balance(self):
    return self.balance

class SavingAccount(Account):
  def __init__(self, balance, rate):
    super().__init__(balance)
    self.rate = float(rate)

  def add_interest(self):
    self.balance += self.balance * self.rate

loai_tk = input().strip()
du_lieu = input().strip().split()

if loai_tk == "Account":
  balance = du_lieu[0]
  acc = Account(balance)

  tien_gui = input().strip()
  acc.deposit(tien_gui)

  print(f'Số dư tài khoản: {acc.get_balance():.2f}')

elif loai_tk == "SavingAccount":
  balance = du_lieu[0]
  rate = du_lieu[1]

  acc_tiet_kiem = SavingAccount(balance, rate)
  acc_tiet_kiem.add_interest()

  print(f'Số dư Sổ tiết kiệm: {acc_tiet_kiem.get_balance():.2f}')
```

```python
class Printer:
    def print_value(self, n):
        print(n)

class BinaryPrinter(Printer):
    def print_value(self, n):
        nhi_phan = bin(n)[2:] # 2: để loại bỏ 0b ở đầu, bin(n) -> đổi sang số nhị phân
        print(nhi_phan)

loai_may = input().strip()
so_n = int(input().strip())

if loai_may == "Printer":
    may_in = Printer()
    may_in.print_value(so_n)

elif loai_may == "BinaryPrinter":
    may_in_np = BinaryPrinter()
    may_in_np.print_value(so_n)
```

### Bài W8A22


**Yêu cầu:** Xây dựng lớp Matrix2x2(a,b,c,d) biểu diễn ma trận. Viết phương thức trace() trả về a+d và determinant() trả về ad-bc.

```python
class Matrix2x2:
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d

    def trace(self):
        return self.a + self.d

    def determinant(self):
        return (self.a * self.d) - (self.b * self.c)

nhap = input().split()
a = float(nhap[0])
b = float(nhap[1])
c = float(nhap[2])
d = float(nhap[3])

mt = Matrix2x2(a, b, c, d)

print(f"{mt.trace():.2f}")
print(f"{mt.determinant():.2f}")
```

### Bài W8A25
**Yêu cầu:** Xây dựng lớp PolynomialSum cho P(x)=ax+b, Q(x)=cx+d. Viết sum_str() trả chuỗi biểu diễn đa thức tổng dạng kx+m.

```python
class PolynomialSum:
    def __init__(self, a, b, c, d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d

    def sum_str(self):
        k = self.a + self.c
        m = self.b + self.d
        if m < 0:
            return f"{k:.2f}x{m:.2f}"
        return f"{k:.2f}x+{m:.2f}"

a, b, c, d = map(float, input().split())
poly_sum = PolynomialSum(a, b, c, d)
print(poly_sum.sum_str())
```

### Bài W8A26
**Yêu cầu:** Xây dựng lớp BaseLogger (log in ra msg) và ErrorLogger ghi đè để in ra: ERROR: msg.

```python
class BaseLogger:
    def __init__(self, msg):
        self.msg = msg

    def log(self):
        print(self.msg)

class ErrorLogger(BaseLogger):
    def __init__(self, msg):
        super().__init__(msg)

    def log(self):
        print(f"ERROR: {self.msg}")

msg = input()
logger = ErrorLogger(msg)
logger.log()
```

### Bài W8A27
**Yêu cầu:** Xây dựng lớp Product(name, price) và DiscountProduct(name, price, percent). Ghi đè final_price() trả về giá sau giảm: price * (100 - percent) / 100.

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def final_price(self):
        return self.price

class DiscountProduct(Product):
    def __init__(self, name, price, percent):
        super().__init__(name, price)
        self.percent = percent

    def final_price(self):
        return self.price * (100 - self.percent) / 100

data = input().split()
name = data[0]
price = float(data[1])
percent = float(data[2])

discount_product = DiscountProduct(name, price, percent)
print(f"{discount_product.final_price():.2f}")
```

### Bài W8A28
**Yêu cầu:** Xây dựng lớp Counter có count ban đầu bằng 0, các phương thức inc(), dec(), value(). Thực hiện dãy lệnh n thao tác INC/DEC và in kết quả.

```python
class Counter:
    def __init__(self):
        self.count = 0

    def inc(self):
        self.count += 1

    def dec(self):
        self.count -= 1

    def value(self):
        return self.count

ops = input().split()
counter = Counter()

for op in ops:
    if op == "INC":
        counter.inc()
    elif op == "DEC":
        counter.dec()

print(f"{counter.value()}")
```

### Bài W8A29
**Yêu cầu:** Xây dựng lớp Time(h, m, s), phương thức to_seconds() đổi thời gian sang tổng số giây.

```python
class Time:
    def __init__(self, h, m, s):
        self.h = h
        self.m = m
        self.s = s

    def to_seconds(self):
        return self.h * 3600 + self.m * 60 + self.s

h, m, s = map(int, input().split())
time_obj = Time(h, m, s)
print(f"{time_obj.to_seconds()}")
```

### Bài W8A30
**Yêu cầu:** Xây dựng lớp Stack bằng danh sách với các phương thức push(x), pop(), top(). Xử lý tập lệnh và mỗi lệnh TOP in giá trị ở đỉnh.

```python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, x):
        self.items.append(x)

    def pop(self):
        self.items.pop()

    def top(self):
        return self.items[-1]

n = int(input())
stack = Stack()

for _ in range(n):
    command = input().split()
    if command[0] == "PUSH":
        stack.push(int(command[1]))
    elif command[0] == "POP":
        stack.pop()
    elif command[0] == "TOP":
        print(f"{stack.top()}")
```
